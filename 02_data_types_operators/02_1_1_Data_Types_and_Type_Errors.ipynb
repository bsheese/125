### Data Type Errors
Let's try to produce a few errors so we can see what they look like. We will only look at a few. You can create many more but this will give you a sense of how to start thinking about the errors.

#### Data Type Error 1

```
# run this cell (click arrow to left or use Ctrl + Enter)
'taco' / 2
```

You should have gotten an error. Specifically, you should see `TypeError: unsupported operand type(s) for /: 'str' and 'int'`.

It may be a little daunting. Let's break it down to see what it means.
* `TypeError` is telling us that the error we made involves some kind of problem with data types.
* `unsupported operand type(s)` means Python is telling us it can't do something (its 'unsupported'). Specifically, it can't do some kind of action or operation on the data types its been given.
* `'str' and 'int'` is telling us that the two specific data types that are involved in this mismatch: a string and an integer.

Okay, so we can't divide the string by an integer (whole number). By design, Python doesn't allow that. What it does allow is multiplying a string by an integer. Fix the code above by replacing the `/` with a `*`.  

#### Data Type Error 2

```
# run this cell (click arrow to left or use Ctrl + Enter)
'1990' - 1.546
```

You should have gotten another error. Specifically, you should see `TypeError: unsupported operand type(s) for -: 'str' and 'float'`.

Again we've got the `TypeError`, and then some details about the problem. In this case, `'1990'` is a string, because of the quotes. Python will not allow us to subtract `1.546` from the string.

This may seem like a very easy error to spot and fix, and it is, but only when we write it this way. Very soon we will be assigning values into variables and it won't be immediately obvious that the value is a string. You won't see `'1990'` in your code, instead it will be a variable, something like `user_response`, that contains a string that will give you this type of error.

Fix the code above so it runs properly.

#### Data Type Error 3

```
# run this cell (click arrow to left or use Ctrl + Enter)
len(2342)
```

You should have gotten an error. Specifically, you should see `TypeError: object of type 'int' has no len()`.

Again we've got the `TypeError`, but what follows is a bit different. In Python, integers don't have a length. So calling the len() function and passing it an integer doesn't work. However, strings do have a length so to fix the function replace the integer with a string, for example `'taco'`. Whatever you put in, don't forget the quotes. If you do, Python will think you've entered a variable name instead of a string.

If you have never had any programming before, feel free to skip this paragraph. If you've had some experience with other languages this error message may seem a bit odd. Notice that Python refers to the integer `2342` as an object and that Python implies that the len() function is not a property of the object. In Python even basic data types are objects with methods attached. In this case, the len function calls a length method of the object.

### Strings
When you are first starting to code you will just think of strings as letters, or words, or text. That's okay to start with, but it's also not really quite right. First off, as we've seen numbers can be strings. Punctuation can be strings. In fact, a quote followed by an empty space and then finished with another quote, like this `' '` is also considered a string. A surprising amount of computer science deals with searching through strings, altering strings, combining strings, and so forth.

What should you know right now? Anything in quotes is a string from Python's perspective. Single quotes (`'`), double quotes (`"`), or even triple quotes (`'''`) don't matter, as long as the string starts and ends with the same type of quote.

### Integers and Floats
Numbers are going to be either ints (integers also known as whole numbers) or floats (floating point numbers, also known as number with decimals). At first, the two will seem almost interchangeable. You'll rarely write code that will throw an error or do something unintended because you handed your code a float instead of an int. But the deeper you get into programming, the more important the distinction will become.  

I do want you to know just a little bit about this at the beginning, because it's so important to programming as a whole. Floating point numbers are more challenging for a computer to deal with than integers. For the computer, everything has to boiled down to binary code (just 0's and 1's); it doesn't deal with the decimal numbers you and I are used to (0-9) and this has some suprising consequences. For example:

```
# make a prediction, then run this cell
0.1 + 0.2
```

The result of the calculation above is incorrect. It highlights the limitations of the computer. Encoding floats in binary doesn't always work like you'd expect. In practical terms, what all this means is that we will favor integers when we can. We won't make things into floats unless we need to. If you'd like to know more, [this video is a good place to start](https://www.youtube.com/watch?v=PZRI1IfStY0).

```
# predict what each line of code will do and then execute this cell
integer_example = 1234
float_example = 1.234

print(integer_example * integer_example)
print(integer_example - float_example)
print(integer_example / float_example)
```

## input()
`input()` is a Python function that asks the user to enter some kind of information. No matter what the user enters, the entered information is returned as a string. If we are asking for text this works just fine. For example...

```
# run this cell then enter your name below
your_name = input('Enter your name: ')
print('Hi,', your_name)
```

But if we are asking for numbers from the user we can't treat the return of `input()` as a number.

```
# run this cell and enter some number of minutes
print('Hi! I convert hours to minutes.')
hours = input('Give me some hours to convert: ')
minutes = hours * 60
print(hours, 'hours is ', minutes, 'minutes')
```

Wow! That really didn't do what we were hoping it would do.

We called `input()` to get hours from the user. Whatever they entered was returned as a string. We then multiplied that string by 60. When Python has a string data type multiplied by some integer it repeats the string that many times. So what we got back was whatever number you entered repeated 60 times.
