## Introduction
When we divide numbers in math, we often expect a decimal result. For example, $7 \div 2 = 3.5$.

However, in programming, we frequently need to perform "Integer Division." This means we only care about how many **whole times** a number goes into another, or what the **remainder** is.

Key concepts covered in this exercise:
1. Floor Division (`//`)
2. Modulus (`%`)

## 1. Floor Division (//)

Imagine you have **7 cookies** and you want to share them equally between **2 friends**. If you aren't allowed to break the cookies into pieces, each friend gets **3 whole cookies**. The fact that there is a "half" cookie left over doesn't matter for this specific question.

In Python, the **Floor Division** operator `//` calculates exactly this: it performs division and "chops off" the decimal part, returning only the whole number (the quotient).

Run the cells below to see it in action.

```
# Standard division returns a float
7 / 2
```

```
# Floor division returns just the whole number
7 // 2
```

## 2. Modulus (%)

Now, let's go back to the **7 cookies** and **2 friends**. We know each friend gets 3 cookies. But how many cookies are **left over** in the jar?

The answer is **1**.

In programming, this "leftover" amount is called the **remainder**. The **Modulus** operator `%` returns only the remainder of a division.

Run the cells below.

```
# 7 divided by 2 is 3 with a remainder of 1
7 % 2
```

```
# 10 divided by 3 is 3 with a remainder of 1
10 % 3
```

## Why use these?

These operators are incredibly useful for tasks like:
*   **Time calculations:** Converting total seconds into minutes and remaining seconds.
*   **Even/Odd checks:** A number is even if `number % 2` is 0.
*   **Grouping:** Figuring out how many full teams you can make and how many people are left out.
