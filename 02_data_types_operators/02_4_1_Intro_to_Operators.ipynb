### Logical Operators

#### Examples
Make a prediction about what the code will produce, then execute each cell.

`and` returns `True` if the code on both sides of the `and` evaluates to `True`. If either returns `False`, then `and` will return `False`.

Make a prediction about what the code will produce, then run each cell below.

```
True and True
```

```
(1 == 1) and (2 == 2)
```

```
True and False
```

```
(1==1) and (2 == 3)
```

`or` returns `True` if either expression returns `True`. It only returns `False` if both sides return `False`.

Make a prediction about what the code will produce, then run each cell below.

```
True or False
```

```
(1==1) or (2==3)
```

```
(1==2) or (2==3)
```

`not` is a little different. It is not a comparison between two values. Instead, it simply reverses a Boolean from `True` to `False` or vice versa.

Make a prediction about what the code will produce, then run each cell below.

```
not (1==1)
```

```
not (1==2)
```

```
not (1==2) or (2 == 3)
```
