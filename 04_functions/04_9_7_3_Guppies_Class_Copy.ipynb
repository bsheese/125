## Introduction

Guppies is a simple betting game where a player starts with 100 dollars and may bet as much of their money as they like every turn. The player places a bet and the computer checks that it is valid. Then the computer selects and announces one random number between 1 and 10. The player guesses if the next random number will be higher or lower than the first number. The second random number is then selected and announced. If the player's guess was correct, then the player earns money equivalent to their bet; otherwise, they lose the money they bet. The game continues until the player loses all of their money or earns 1000 dollars.

Write a program that plays guppies.

Your game must:
- import the random library with `import random`
- print a welcome message
- report your current money (start with $100)
- ask for your bet
- check if your bet is valid, and continue to ask until it is
- use `random.randint(1,10)` to generate and print a random integer between 1 and 10
- ask the user if a second number will be higher or lower
- generate and print a second random number
- check the players guess, add money if they won, subtract if they lost
- end the game if the player reachers either 0 or 1000, otherwise continue
- after the game ends print the outcome (win or loss) 

```
# you don't have to wait, but we will start coding guppies in class

# step 1 import random library
import random

# step 2 print welcome messauge
print('Welcome to Guppies!')
print()

# step 3 initialize money
money = 100 


# step 4 start game loop
while (money > 0) and (money < 1000):

  # step 5 report current money
  print('You currently have:', money) 


  # step 6 ask for money then check if valid
  while True:
    # ask for your bet
    bet = int(input('How much do you want to bet? '))
    
    if bet > money:
      print('You cannot bet more money than you have!')
    else:
      break


  # step 7 select and print random integer
  rnumber1 = random.randint(1,10)
  print('The first number is ', rnumber1)


  # step 8 get guess higher or lower
  guess = input('Is your guess (h)igher or (l)ower?') 

  # step 9 select and print second random integer
  rnumber2 = random.randint(1,10)
  print('The second number is ', rnumber2)

  # step 10 check guess and update money
  if guess == 'h':
    if rnumber2 > rnumber1:
      print('You were right!')
      money = money + bet
    else: 
      print('You were wrong!')
      money = money - bet

  if guess == 'l':
    if rnumber2 < rnumber1:
      print('You were right!')
      money = money + bet
    else: 
      print('You were wrong!')
      money = money - bet  

if money < 1:
  print('You lost!')
else:
  print('You won!')
  
print('Your final money is: ', money)
print('Game over!')
```
