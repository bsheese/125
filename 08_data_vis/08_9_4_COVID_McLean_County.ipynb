# Introduction
The Center for Disease Control has provided county level data on reported COVID-19 cases. We have a subset of that data here for McLean County Illinois that covers March of 2020 through March of 2021.

```python
# imports
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# set figure size
sns.set(rc={'figure.figsize':(22,8)})

# get the data and constuct the data frame
url = 'https://raw.githubusercontent.com/'
repo = 'bsheese/CSDS125ExampleData/master/'
fnme = 'data_covid_mclean_county_retrieved_4_1_21.csv'
df = pd.read_csv(url + repo + fnme)

# convert a series to datatime format
# weird temporary hack to avoid silent fail or dt conversion
df.loc[:, 'case_month1'] = pd.to_datetime(df.loc[:, 'case_month'], format='%Y-%m')
df = df.drop('case_month', axis = 1)
df = df.rename({'case_month1': 'case_month'}, axis=1)
# check dataframe
df.head(3)
```

## Using Datetime Datatype
The datetime datatype allows us to represent information about dates and time in a variety of ways and access differnet parts of a date or time representation easily. For example:

```python
# examining the datetime series
df.loc[:, 'case_month'].head()
```

```python
# using datetime accessor w series w month and year encoded as year-month
df.loc[:, 'case_month'] \
  .dt \
  .year
```

```python
# using datetime accessor w series w month and year encoded as year-month
df.loc[:, 'case_month'] \
  .dt \
  .month
```
