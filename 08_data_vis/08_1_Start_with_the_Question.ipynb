# 08.1: Start with the Question

A table of numbers can answer a specific question, but it is hard to see the shape of a column, or a pattern across many rows, by reading a table. A chart can show those things at once. The difficulty is that a chart is only useful if it fits the question, and Python will draw almost anything you ask for whether or not it makes sense. So this module does more than show how to draw charts. For each kind of chart, it covers when to choose it, what can go wrong with it and how to read it.

This notebook introduces the two questions to ask before you draw anything, and the code pattern that nearly every chart in this module uses. The later notebooks each take up one kind of question in detail.

## Two questions before you draw

**What am I asking?** Most questions about a table are one of five kinds: comparing categories, describing one numeric column, comparing a numeric column across groups, looking for a relationship between two numeric columns, or following something over time.

**What kinds of columns do I have?** The same four kinds of column come up again and again:

| Kind of column | What it holds | Example |
|---|---|---|
| Categories | Labels with no natural order | species, island |
| Ordered categories | Labels with an order | small, medium, large |
| Numbers | Amounts you can do math on | body mass |
| Time | Moments in order | year, month |

The kinds of columns matter because a chart turns values into something you can see, such as the position of a point or the length of a bar. Position and length both suggest an amount or an order, so they can suggest something false if the column is not really like that. A number stored in a column is not always an amount, either: a ZIP code is a label, even though it is written with digits.

## The data

We start with measurements of penguins from three species in Antarctica. Before you draw anything from a table, you need to know what one row represents, as you saw in module 07. Here one row is one penguin.

`import seaborn as sns` loads **seaborn**, a Python library for making charts, under its usual short name. Seaborn is built on a library called **matplotlib**, which we load as `plt`. We use `sns` to draw and matplotlib for small adjustments.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

penguins = pd.read_csv("https://raw.githubusercontent.com/bsheese/125/08-rebuild/data/data_penguins.csv")
print(penguins.shape)
penguins.head(3)
```

There are 344 penguins and 7 columns. `species`, `island` and `sex` are categories, and the columns ending in `_mm` and `_g` are numbers. No column records time, so this table cannot answer a question about change over time.

## The pattern

Nearly every chart in this module follows one pattern. Tell seaborn which DataFrame to use with `data=` and which columns to use with `x=` and `y=`. Seaborn returns an object that we store as `ax`, short for axes, which represents the chart. We use `ax` to add a title and axis labels, and `plt.show()` displays the result.

Here is the first question: how many penguins of each species are in this data?

```python
ax = sns.countplot(data=penguins, x="species")
ax.set_title("Number of penguins by species")
ax.set_xlabel("Species")
ax.set_ylabel("Number of penguins")
plt.show()
```

There are 152 Adelie, 124 Gentoo and 68 Chinstrap penguins. `sns.countplot()` counts the rows in each category for you, so we gave it only `x=`. This is a **bar chart**, and it fits because the question compares categories.

## Same table, different questions

The next three questions use the same table, but each one calls for a different chart. Notice that the choice follows from the question and the kinds of columns, not from the data alone.

**What does one numeric column look like?** A **histogram** divides a numeric column into ranges and shows how many values fall in each.

```python
ax = sns.histplot(data=penguins, x="body_mass_g")
ax.set_title("Body mass of penguins")
ax.set_xlabel("Body mass (grams)")
ax.set_ylabel("Number of penguins")
plt.show()
```

Most penguins weigh between about 3000 and 5000 grams, with a peak near 3800 and a longer tail toward the heavy end. The chart shows the whole shape of the column, which a single average could not. Notebook 08.3 covers histograms.

**How does a numeric column differ across groups?** A **box plot** summarizes the column once for each group, so that the groups can be lined up and compared.

```python
ax = sns.boxplot(data=penguins, x="species", y="body_mass_g")
ax.set_title("Body mass by species")
ax.set_xlabel("Species")
ax.set_ylabel("Body mass (grams)")
plt.show()
```

Gentoo penguins are clearly heavier, with a median near 5000 grams, while Adelie and Chinstrap penguins are similar to each other, with medians near 3700. That explains the long tail in the histogram: the heavy end is mostly Gentoo penguins. Notebook 08.4 covers box plots.

**Are two numeric columns related?** A **scatter plot** draws one point per row, with one column on each axis.

```python
ax = sns.scatterplot(data=penguins, x="flipper_length_mm", y="body_mass_g")
ax.set_title("Flipper length and body mass")
ax.set_xlabel("Flipper length (mm)")
ax.set_ylabel("Body mass (grams)")
plt.show()
```

The points rise from left to right, so penguins with longer flippers tend to be heavier. Notebook 08.5 covers scatter plots.

**How does something change over time?** The penguin table cannot answer this, so we need another one. In this table one row is one month of airline passengers, from 1949 to 1960. `parse_dates=` tells pandas to treat the `date` column as dates. A **line plot** connects the months in order.

```python
flights = pd.read_csv("https://raw.githubusercontent.com/bsheese/125/08-rebuild/data/data_flights.csv", parse_dates=["date"])
ax = sns.lineplot(data=flights, x="date", y="passengers")
ax.set_title("Monthly airline passengers, 1949-1960")
ax.set_xlabel("Month")
ax.set_ylabel("Passengers (thousands)")
plt.show()
```

The line climbs from 112 thousand passengers in January 1949 to 432 thousand in December 1960, with a repeating rise and fall within each year. A line fits because the months have a real order and the points between them are connected in time. Notebook 08.6 covers line plots and when to use a bar chart instead.

## Python will not stop you

Here is a chart that runs without any error and tells you almost nothing. It puts one category on each axis.

```python
ax = sns.scatterplot(data=penguins, x="species", y="island")
ax.set_title("A chart that does not fit its columns")
ax.set_xlabel("Species")
ax.set_ylabel("Island")
plt.show()
```

The 344 penguins pile up on just five points, so you cannot see how many are at each one. A scatter plot needs numbers, because it depends on the points spreading out. The question this chart seems to ask, which species live on which islands, has a better answer: `pd.crosstab()` from notebook 07.4 gives the counts directly. Python's job is to draw what you ask for, so checking that the chart fits the question is up to you.

## Which chart for which question?

This table is a starting point, and the later notebooks explain the exceptions.

| Question | Kinds of columns | Chart | Function |
|---|---|---|---|
| How do categories compare? | One category, one number or a count | Bar chart | `sns.barplot`, `sns.countplot` |
| What does one numeric column look like? | One number | Histogram | `sns.histplot` |
| How does a number differ across groups? | One category, one number | Box plot | `sns.boxplot` |
| Are two numeric columns related? | Two numbers | Scatter plot | `sns.scatterplot` |
| How does something change over time? | Time, one number | Line plot | `sns.lineplot` |

## A title and labels are part of the chart

Every chart above has a title and labeled axes, and the labels include units such as grams or thousands of passengers. Seaborn would otherwise label the axes with the column names, and a name such as `body_mass_g` means little to someone who has not seen your table. It is also part of the job to say what a chart shows. A useful description names the pattern and gives numbers, as in "Gentoo penguins are heavier, with a median near 5000 grams, compared with about 3700 for the other two species." The rest of this module asks you to do this after each chart.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

cars = pd.read_csv("https://raw.githubusercontent.com/bsheese/125/08-rebuild/data/data_mpg.csv")
cars_start = cars.copy()
origin_counts = cars["origin"].value_counts().sort_values()
```

```python
#@title Run this cell to set up the checks
import numpy as np
import pandas as pd


def same(got, want):
    """True when got matches want: same labels and values for a series, same value for a number."""
    try:
        if isinstance(want, pd.DataFrame):
            if not isinstance(got, pd.DataFrame) or list(got.columns) != list(want.columns) or len(got) != len(want):
                return False
            if not pd.api.types.is_integer_dtype(want.index) and list(got.index) != list(want.index):
                return False
            got, want = got.reset_index(drop=True), want.reset_index(drop=True)
            for column in want.columns:
                if pd.api.types.is_numeric_dtype(want[column]):
                    if not np.allclose(pd.to_numeric(got[column], errors="coerce").astype(float), want[column].astype(float), equal_nan=True):
                        return False
                elif got[column].fillna("").astype(str).to_list() != want[column].fillna("").astype(str).to_list():
                    return False
            return True
        if isinstance(want, pd.Series):
            if not isinstance(got, pd.Series) or list(got.index) != list(want.index):
                return False
            if pd.api.types.is_numeric_dtype(want):
                return bool(np.allclose(got.astype(float), want.astype(float), equal_nan=True))
            return list(got) == list(want)
        if isinstance(want, (list, np.ndarray)):
            return list(got) == list(want)
        if isinstance(want, str):
            return got == want
        if isinstance(got, (pd.Series, pd.DataFrame, list, np.ndarray)):
            return False
        return bool(np.isclose(float(got), float(want)))
    except (TypeError, ValueError):
        return False


def check(number, name, want, hint, unchanged=None):
    """Print feedback for one problem. unchanged is (variable name, its starting value)."""
    if name not in globals():
        print(f"Problem {number}: not yet. No variable named {name} exists. Did you run your answer cell?")
    elif not same(globals()[name], want):
        print(f"Problem {number}: not yet. {hint}")
    elif unchanged is not None and not globals()[unchanged[0]].equals(unchanged[1]):
        print(f"Problem {number}: {name} is right, but {unchanged[0]} changed. Work on a copy, then run the setup cell again.")
    else:
        print(f"Problem {number}: correct.")


def check_chart(number, name, test, hint, y_label=True):
    """Check a chart stored in a variable: it has labels, and test(chart) is True."""
    if name not in globals():
        print(f"Problem {number}: not yet. No variable named {name} exists. Did you run your answer cell?")
        return
    chart = globals()[name]
    if not hasattr(chart, "get_title"):
        print(f"Problem {number}: not yet. Store what the seaborn function returns in {name}.")
    elif not chart.get_title() or not chart.get_xlabel() or (y_label and not chart.get_ylabel()):
        print(f"Problem {number}: not yet. Give the chart a title and label its axes.")
    else:
        try:
            ok = bool(test(chart))
        except (AttributeError, IndexError, TypeError, ValueError):
            ok = False
        print(f"Problem {number}: correct." if ok else f"Problem {number}: not yet. {hint}")
```
