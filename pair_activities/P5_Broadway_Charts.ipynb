## Setup
Run this cell first.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("https://raw.githubusercontent.com/bsheese/CSDS125ExampleData/master/data_corgis_broadway.csv")
df.head(3)
```

## Q1

```python
# Q1
ax = sns.histplot(data=df, x="Attendance")
ax.set_title("Weekly attendance")
ax.set_xlabel("People attending in a week")
ax.set_ylabel("Number of show-weeks")
plt.show()
print(df["Attendance"].describe())
```

## Q2

```python
# Q2
ax = sns.scatterplot(data=df, x="Year", y="Month")
plt.show()
```

## Q3

```python
# Q3
top = df.groupby("Name")["Gross"].sum().sort_values(ascending=False).head(10).reset_index()
top["Billions"] = top["Gross"] / 1000000000
ax = sns.barplot(data=top, x="Name", y="Billions")
plt.show()
```

## Q4

```python
# Q4
by_year = df.groupby("Year")["Gross"].sum().reset_index()
by_year["Billions"] = by_year["Gross"] / 1000000000
ax = sns.lineplot(data=by_year, x="Year", y="Billions")
ax.set_title("Total Broadway gross by year")
ax.set_xlabel("Year")
ax.set_ylabel("Total gross (billions of dollars)")
plt.show()
print(by_year[by_year["Year"].isin([1992, 2015, 2016])])
```

## Q5

```python
# Q5
rows = df.groupby("Year")["Gross"].count().reset_index()
rows = rows.rename(columns={"Gross": "Rows"})
rows.head(8)
```

## Q6

```python
# Q6
steady = by_year[(by_year["Year"] >= 1998) & (by_year["Year"] <= 2015)]
ax = sns.lineplot(data=steady, x="Year", y="Billions")
ax.set_title("Total Broadway gross, 1998-2015")
ax.set_xlabel("Year")
ax.set_ylabel("Total gross (billions of dollars)")
plt.show()
```
