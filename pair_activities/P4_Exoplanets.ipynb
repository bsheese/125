# Pair Activity 4: Exoplanets
The quiz on Canvas tells you what to do in each section. This data lists planets found outside our solar system through 2014, with the method used to find each one.

## Setup
Run this cell first.

```python
import pandas as pd

planets = pd.read_csv("https://raw.githubusercontent.com/mwaskom/seaborn-data/master/planets.csv")
planets.head()
```

## Q1

```python
# Q1
print(len(planets))
planets.isna().sum()
```

## Q2

```python
# Q2
print(len(planets.dropna()))
print(len(planets.dropna(subset=["distance"])))
```

## Q3

```python
# Q3
print(planets["mass"].mean())
print(planets["mass"].fillna(0).mean())
```

## Q4

```python
# Q4
planets.groupby("method")["distance"].mean().sort_values(ascending=False)
```

## Q5

```python
# Q5
planets.groupby("method")["distance"].count().sort_values(ascending=False)
```

## Q6

```python
# Q6
planets["Period"] = pd.cut(planets["year"], bins=[1988, 2004, 2009, 2014], labels=["through 2004", "2005-2009", "2010-2014"])
two_methods = planets[planets["method"].isin(["Radial Velocity", "Transit"])]
pd.crosstab(two_methods["Period"], two_methods["method"])
```
