## Setup
Run this cell first. It makes three series with the same labels: `violent` (violent crime rate), `property_crime` (property crime rate) and `population`.

```python
import pandas as pd

crime = pd.read_csv("https://raw.githubusercontent.com/bsheese/CSDS125ExampleData/master/data_corgis_state_crime.csv")
crime = crime[crime["Year"] == 2012]
labels = crime["State"].to_list()
violent = pd.Series(crime["Rates.Violent.All"].to_list(), index=labels)
property_crime = pd.Series(crime["Rates.Property.All"].to_list(), index=labels)
population = pd.Series(crime["Population"].to_list(), index=labels)
violent.head()
```
