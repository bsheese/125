# Pair Activity 2: Crime and Oscars
The quiz on Canvas tells you what to do in each section. Part 1 uses the 2012 crime rates from Pair Activity 1, and Part 2 uses the Best Actor and Best Actress Oscar winners.

## Setup
Run this cell first.

```python
import pandas as pd

crime = pd.read_csv("https://raw.githubusercontent.com/bsheese/CSDS125ExampleData/master/data_corgis_state_crime.csv")
crime = crime[crime["Year"] == 2012]
violent = pd.Series(crime["Rates.Violent.All"].to_list(), index=crime["State"].to_list())

oscars = pd.read_csv("https://raw.githubusercontent.com/bsheese/CSDS125ExampleData/master/data_oscars.csv")
winners = pd.Series((oscars["name"] + " in " + oscars["movie"]).to_list(), index=oscars["oscar_yr"].to_list())
winners.head(3)
```

```python
# Q2
states = violent[~violent.index.isin(["United States", "District of Columbia"])]
states.describe()
```
