# Pair Activity: Crime and Oscars
The quiz on Canvas tells you what to do in each section. Part 1 uses the 2012 crime rates from the State Crime pair activity, given as violent crimes per 100,000 people, and Part 2 uses the Best Actor and Best Actress Oscar winners.

## Setup
Run this cell first.

```python
import pandas as pd

crime = pd.read_csv("https://raw.githubusercontent.com/bsheese/CSDS125ExampleData/master/data_corgis_state_crime.csv")
crime = crime[crime["Year"] == 2012]
violent = pd.Series(crime["Rates.Violent.All"].to_list(), index=crime["State"].to_list())

oscars = pd.read_csv("https://raw.githubusercontent.com/bsheese/CSDS125ExampleData/master/data_oscars.csv")
winners = pd.Series((oscars["name"] + " in " + oscars["movie"]).to_list(), index=oscars["oscar_yr"].to_list())
winners.head(3)
```

## Q1

```python
# Q1
violent.describe()
```

## Q2

```python
# Q2: removes only D.C. United States stays in for now, and Q3 removes it.
no_dc = violent[~violent.index.isin(["District of Columbia"])]
no_dc.describe()
```

## Q3

```python
# Q3: keeps only the 50 states
states = violent[~violent.index.isin(["United States", "District of Columbia"])]
print("25th percentile:", states.quantile(0.25))
print("75th percentile:", states.quantile(0.75))
print("Illinois:", violent.loc["Illinois"])
```
