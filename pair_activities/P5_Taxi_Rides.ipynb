# Pair Activity 5: Taxi Rides
The quiz on Canvas tells you what to do in each section. This data lists a sample of New York City taxi rides from March 2019, where one row is one ride.

## Setup
Run this cell first.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

taxis = pd.read_csv("https://raw.githubusercontent.com/bsheese/125/main/data/data_taxis.csv")
taxis.head(3)
```

## Q1

```python
# Q1
ax = sns.countplot(data=taxis, x="pickup_borough")
ax.set_title("Taxi rides by pickup borough")
ax.set_xlabel("Pickup borough")
ax.set_ylabel("Number of rides")
plt.show()
```

## Q2

```python
# Q2
ax = sns.histplot(data=taxis, x="fare")
ax.set_title("Taxi fares")
ax.set_xlabel("Fare (dollars)")
ax.set_ylabel("Number of rides")
plt.show()
print(taxis["fare"].describe())
```

## Q3

```python
# Q3
under_80 = taxis[taxis["fare"] < 80]
print(len(taxis) - len(under_80))
ax = sns.histplot(data=under_80, x="fare", bins=10)
ax.set_title("Taxi fares under 80 dollars")
ax.set_xlabel("Fare (dollars)")
ax.set_ylabel("Number of rides")
plt.show()
```

## Q4

```python
# Q4
mean_tip = taxis.groupby("payment")["tip"].mean().reset_index()
ax = sns.barplot(data=mean_tip, x="payment", y="tip")
ax.set_title("Mean tip by payment type")
ax.set_xlabel("Payment type")
ax.set_ylabel("Mean tip (dollars)")
plt.show()
print(mean_tip)
```

## Q5

```python
# Q5
median_order = taxis.groupby("pickup_borough")["fare"].median().sort_values(ascending=False).index
ax = sns.boxplot(data=taxis, x="fare", y="pickup_borough", order=median_order)
ax.set_title("Fares by pickup borough")
ax.set_xlabel("Fare (dollars)")
ax.set_ylabel("")
plt.show()
print(taxis.groupby("pickup_borough")["fare"].median())
```

## Q6

```python
# Q6
queens = taxis[taxis["pickup_borough"] == "Queens"]
ax = sns.histplot(data=queens, x="fare", binwidth=5)
ax.set_title("Fares for rides picked up in Queens")
ax.set_xlabel("Fare (dollars)")
ax.set_ylabel("Number of rides")
plt.show()
print(queens["pickup_zone"].value_counts().head(3))
```
