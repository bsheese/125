## Quick Reference
### List Indexing, Slicing
- list_a[1] # index single value
- list_a[1:3] # index multiple values
- list_a[1:10:10] # index values with step 
- list_a[:] # copy of the whole list 

### List Operations
 - list_c = list_a + list_b
 - list_multipled = list_a * 3

### Adding to a List
- list_a.append(item_a) # append item_a to a list
- list_a.extend(list_b) # extend list_a by adding elements from list_b

### Sorting a List
- list_a.sort() # sorts the list in place, returns None

### Removing from a List
- list_a.pop(1) # remove the item at index 1, returns the item
- list_a.remove('dog') # remove the item with value 'dog', returns None

- del list_a[1] # the del operator also works, but please stick with pop() for now

### Functions that can take lists as arguments
- len(), max(), min(), sum()

### Changing objects into lists
- list(string1) # returns new list with individual characters from string1 as values
- string1.split(',') # returns new list with characters from string1 split up into groups using the ',' character

### Changing lists into strings
- list_a.join(',') # returns a string where all values in list_a are joined together by a comma
