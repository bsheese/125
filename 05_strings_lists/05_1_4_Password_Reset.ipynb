## A selection of string methods

### `.isdigit()`

````
# returns true if the string only contains digits
'123'.isdigit()
````

````
'a123'.isdigit()
````

### `.isalpha()`

````
# returns True if the string only contain alphabetic characters
'abc'.isalpha()
````

````
'abc123'.isalpha()
````

### `.isalnum()`

````
# cat only contains alphanumeric characters (letters and numbers)
"cat123".isalnum()
````

````
"cat123!".isalnum()
````

### `islower()`

````
# cat only contain lower case characters
'cat'.islower()
````

````
'Cat'.islower()
````

### `isupper()`

````
# cat only contains uppercase characters
'CAT'.isupper()
````

````
'Cat'.isupper()
````
