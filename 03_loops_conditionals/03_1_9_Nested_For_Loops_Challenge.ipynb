## Nested For Loops
Nested for loops involve loops that are called inside other loops. For example:

````
# nested loop example 1
for i in range(3):
    print('This is the first level loop. Iteration:', i)
    for i2 in range(3):
        print('     This the second level loop. Iteration:', i2)
````

````
# nested loop example 2
for i in range(3):
  for i2 in range(3):
    print(i, i2)
````

````
# nested loop example 3
for i_in_first in range(6):
    for i_in_second in range(i_in_first+1):
        print(i_in_second, end = '')
    print()
````
