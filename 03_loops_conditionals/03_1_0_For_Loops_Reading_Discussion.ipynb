Briefly explain loops and lists. Give an example of how a list can be used in a loop.

```python
for item in range(15000):
  print(item)
```

```python
for i in range(10):
```

```python
for x in 'dogdogdog12343':
  print('mushroom')
  print('cat')
```

```python
# example of a for loop

# for item in some sequence (in this case a list)
for item in ['A', 'B', 'C']:

  # do something for each item in the sequence
  print(item)
```

```python
# example of list
list1 = [1, 2, 3, 4, 5]

# example of for loop using a list
for item in list1:
  print(list1)




```
