A counted loop is loop that keeps track of how many time a loop has run.
You start a count with zero before the loop, and then just add one to the count inside the loop. After the loop has finished the count will equal the total number of times the loop has run. Here is an example:

```
count = 0
for letter in 'deoxyribonucleic acid':
  count = count + 1

print(count)
```
