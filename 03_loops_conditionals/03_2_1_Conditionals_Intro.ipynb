## Problem 11
Fix the code below so it prints:
```
It is 55 degrees today. Stay Inside!
It is 65 degrees today. Just right!
It is 95 degrees today. Stay Inside!
It is 35 degrees today. Stay Inside!
It is 75 degrees today. Just right!


```

````
temperatures = [55, 65, 95, 35, 75]

for temperature in temperatures:
  if temperature > 80 and temperature < 60:
    warning = "Stay Inside!"
  else:
    warning = 'Just right!'
    
  print(f'It is {temperature} degrees today. {warning}')
````

## Problem 12
Fix the code below so it prints:
```
It is 55 degrees today. Too cold!
It is 65 degrees today. Just right!
It is 95 degrees today. Too hot!
It is 35 degrees today. Too cold!
It is 75 degrees today. Just right!
```

````
temperatures = [55, 65, 95, 35, 75]

for temperature in temperatures:
  warning = 'Just right!' # do not edit, move, or copy this line
  # you may only edit the code below
  if temperature < 60:
    warning = "Too cold!"
  if temperature < 80:
    warning = "Too hot!"  
    
  print(f'It is {temperature} degrees today. {warning}')
````
